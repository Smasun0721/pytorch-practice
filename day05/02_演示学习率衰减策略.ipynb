```python
from statistics import mean

import torch
from torch import nn, optim
from matplotlib import pyplot as plt
```

```python
# 1.定义变量
lr = 0.1  # 学习率
epochs = 200  # 训练轮数
iteration = 10  # 每轮批次数

# 2.创建数据集
# 真实值
y_true = torch.tensor([3, 2, 5])
# 输入值
x = torch.tensor([3, 2, 1], dtype=torch.float32)
# 权重
w = torch.tensor([1, 2, 3], requires_grad=True, dtype=torch.float32)

# 3.创建优化器对象
optimizer = optim.SGD([w], lr=lr, momentum=0.9)  # 动量法

# 4.创建学习率衰减对象
scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=50, gamma=0.5)
# 创建损失函数对象
criterion = nn.MSELoss()

# 5.创建列表记录学习率与轮数变化
lr_list = []
epochs_list = []

# 6.训练
for epoch in range(epochs):
    # 记录数据
    lr_list.append(sum(scheduler.get_last_lr()) / len(scheduler.get_last_lr()))
    epochs_list.append(epoch+1)

    # 批次训练
    for batch in range(iteration):
        # 梯度清零
        optimizer.zero_grad()
        # 计算损失
        y_pred = w * x
        loss = criterion(y_pred, y_true).sum()
        # 反向传播
        loss.backward()
        # 优化器更新
        optimizer.step()

    # 更新学习率
    scheduler.step()

# 可视化数据
plt.plot(epochs_list, lr_list)
plt.show()

```
