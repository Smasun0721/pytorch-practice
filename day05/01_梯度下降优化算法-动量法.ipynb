```python
import torch
from torch import nn
from torch import optim
from matplotlib import pyplot as plt
```

```python
# 初始化权重参数
w = torch.tensor([1., 2., 3.], requires_grad=True)
# 初始化优化器
optimizer = optim.SGD([w], lr=0.01, momentum=0.9)
# 定义损失函数
criterion = w ** 2 / 20

loss_list = []
# 循环优化更新
for _ in range(100):
    # 梯度清零
    optimizer.zero_grad()
    # 计算损失函数
    loss = (w ** 2 / 20).sum()
    loss_list.append(loss.item())
    # 反向传播
    loss.backward()
    # 更新参数
    optimizer.step()

    # 打印数据
    print(f'w: {w}\nw.grad: {w.grad}')
    print(f'criterion: {loss}\n')

print(loss_list)
plt.plot(loss_list)
plt.show()
```
