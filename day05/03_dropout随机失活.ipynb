```python
import torch
from torch import nn
```

```python
# 1.创建输入层数据
t1=torch.randint(1,10,size=(1,4),dtype=torch.float)
print(t1)

# 2.创建全连接层
linear1 = nn.Linear(4, 5)

# 3.加权求和
l1 = linear1(t1)
print(l1.detach().numpy())

# 4.激活函数
output = torch.relu(l1)
print(output)

# 5.dropout随机失活
dropout = nn.Dropout(p=0.5)
d1 = dropout(output)
print(d1)
```
