```python
import torch
import torch.nn as nn
```

```python
y_true = torch.tensor([1, 2, 2], dtype=torch.float32)
y_pred = torch.tensor([3, 4, 5], requires_grad=True, dtype=torch.float32)

criterion = nn.L1Loss()
loss = criterion(y_pred, y_true)
print(f'MAELoss: {loss}')
```

```python
criterion = nn.MSELoss()
loss = criterion(y_pred, y_true)
print(f'MSELoss: {loss}')
```

```python
criterion = nn.SmoothL1Loss()
loss = criterion(y_pred, y_true)
print(f'SmoothL1Loss: {loss}')
```
