```python
import torch
from torch import nn
from matplotlib import pyplot as plt
```

```python
img = torch.tensor(plt.imread('./data/img.jpg'), dtype=torch.float32)
conv = nn.Conv2d(in_channels=3, out_channels=4, kernel_size=3)
# HWC -> CHW
img = torch.permute(img, (2, 0, 1))
batch = img.unsqueeze(0)
print(f'batch.shape: {batch.shape}')
# 卷积层
output = conv(batch)[0]
print(f'output.shape: {output.shape}')
# 池化层
output = torch.max_pool2d(output, kernel_size=3, stride=2)
print(f'output.shape: {output.shape}')
# CHW -> HWC
output = torch.permute(output, (1, 2, 0))
print(f'output.shape: {output.shape}')
for i in range(output.shape[-1]):
    feature = output[:, :, i].detach().numpy()
    print(f'feature{i}.shape: {feature.shape}')
    plt.imshow(feature)
    plt.show()
```
